## Task 1 - Write Sales Records to a File

### Create a list of sales amounts:
###  sales = [1200, 450, 980, 1500, 3000]
###  Open a file named sales_data.txt in write mode.
###  Write each sale on a new line in the file
###  Close the file, then reopen it and print its content.
### Write the data in comma-separated format instead of separate lines.

```python
sales = [1200, 450, 980, 1500, 3000]
```

```python
# Create a list of sales amounts
sales = [1200, 450, 980, 1500, 3000]

# Open the file in write mode and write comma-separated values
file = open("sales_data.txt", "w")
file.write(",".join(map(str, sales)))
file.close()

# Reopen the file in read mode and print its content
file = open("sales_data.txt", "r")
content = file.read()
print(content)
file.close()
```

## Task 2 - Read file in different ways

### Using the same sale_data.txt
### 1. Read the entire file using .read() and print it
### 2. Read the first line using .readline()
### 3. Read the lines using .readline() and convert them into a list of integers
### Esure proper  formatting and cleanup of newline characters

```python
# 1. Read the entire file using read()
file = open("sales_data.txt", "r")
content = file.read()
print("1. Entire file:")
print(content)
file.close()


# 2. Read the first line using readline()
file = open("sales_data.txt", "r")
first_line = file.readline()
print("\n2. First line:")
print(first_line.strip())
file.close()


# 3. Read the line and convert values into a list of integers
file = open("sales_data.txt", "r")
line = file.readline()

# Remove newline characters and split using commas
sales_list = [int(value.strip()) for value in line.strip().split(",")]

print("\n3. Sales as a list of integers:")
print(sales_list)

file.close()
```

## Task 3 - Append New Sales

### 1. Append these new sales to the same file: 5000, 2500, 1700
### 2. After appending , reopen  and print the entire updated file.
### (Extra) print total number of lines after appending.

```python
# New sales amounts
new_sales = [5000, 2500, 1700]

# Open the existing file in append mode
file = open("sales_data.txt", "a")
file.write("," + ",".join(map(str, new_sales)))
file.close()

# Reopen the file in read mode
file = open("sales_data.txt", "r")
content = file.read()
print("Updated sales data:")
print(content)
file.close()

# Count the total number of lines
file = open("sales_data.txt", "r")
lines = file.readlines()
print("Total number of lines:", len(lines))
file.close()
```

## Task 4 - Generate Summary Report from File

### Using only files reading operations:
### 1. Read all sales values from sales_data.txt
### 2. Convert them into integers.
### 3. Calculate and print:
### Total sale
### Highest sale
### Lowest sale
### Average sale
### Do not use any advanced libraries

```python
# Open the file in read mode
file = open("sales_data.txt", "r")

# Read all sales values
content = file.read()

# Close the file
file.close()

# Split comma-separated values and convert them into integers
sales = [int(value) for value in content.split(",")]

# Calculate summary statistics
total_sale = sum(sales)
highest_sale = max(sales)
lowest_sale = min(sales)
average_sale = total_sale / len(sales)

# Print the summary report
print("Sales Summary Report")
print("--------------------")
print("Total Sale:", total_sale)
print("Highest Sale:", highest_sale)
print("Lowest Sale:", lowest_sale)
print("Average Sale:", average_sale)
```

## Task 5 - Create Product info File (User Input)

### 1. Ask the user 3 product names & their prices.
### 2. Write them into a new file products.txt in this format.
### 3. ProductName | Price
### 4. Read the file and print each line with proper formatting.

```python
# Open the file in write mode
file = open("products.txt", "w")

# Ask the user for 3 product names and prices
for i in range(3):
    product_name = input(f"Enter product {i + 1} name: ")
    price = input(f"Enter product {i + 1} price: ")

    # Write product details into the file
    file.write(f"{product_name} | {price}\n")

# Close the file
file.close()

# Reopen the file in read mode
file = open("products.txt", "r")

# Read and print each line with proper formatting
print("\n--- Product Information ---")

for line in file:
    print(line.strip())

# Close the file
file.close()
```

## Task 6 - Read file Safely (Error Handling Inside File Handling Only)

### You must not use expectations beyond file-related safegaurds here.
### 1. Ask the user for a filename to open.
### 2. If the file exists, read and print it.
### 3. If it does not exist, print
### "File not found. Please check the filename."
### Use simple condition checks with os.path.exists() (allowed).

```python
import os

# Ask the user for the filename
filename = input("Enter the filename: ")

# Check whether the file exists
if os.path.exists(filename):
    # Open the file in read mode
    file = open(filename, "r")

    # Read and print the content
    content = file.read()
    print(content)

    # Close the file
    file.close()

else:
    print("File not found. Please check the filename.")
```

## Task 7 - Mini Project - Export Discounted Price

### Create a dictionary :
### price = { "mouse": 500, "Keyboard": 800, "Monitor": 7000, "Pendrive": 400, "Camera": 5000 }
### Ask the user for a discount percentage
### Write discounted price into discount_report.txt using:
### Product | Original Price | Discount Price.
### After writing, read the file and print it to the terminal
### Extra (optional): Write a summary at the bottom of the file:
### Total Items: X
### Average Discounted Price: Y

```python
# Create a dictionary of product prices
price = {
    "mouse": 500,
    "Keyboard": 800,
    "Monitor": 7000,
    "Pendrive": 400,
    "Camera": 5000
}

# Ask the user for discount percentage
discount = float(input("Enter discount percentage: "))

# Initialize variables
total_discounted_price = 0
total_items = 0

# Open the file in write mode
file = open("discount_report.txt", "w")

# Write the heading
file.write("Product | Original Price | Discount Price\n")
file.write("------------------------------------------\n")

# Calculate discounted prices
for product, original_price in price.items():

    discounted_price = original_price - (original_price * discount / 100)

    # Write product details into the file
    file.write(f"{product} | {original_price} | {discounted_price:.2f}\n")

    total_discounted_price += discounted_price
    total_items += 1

# Calculate average discounted price
average_price = total_discounted_price / total_items

# Write summary at the bottom
file.write("------------------------------------------\n")
file.write(f"Total Items: {total_items}\n")
file.write(f"Average Discounted Price: {average_price:.2f}\n")

# Close the file
file.close()

# Reopen the file in read mode
file = open("discount_report.txt", "r")

# Read and print the content
content = file.read()
print("\nDiscount Report:")
print(content)

# Close the file
file.close()
```
